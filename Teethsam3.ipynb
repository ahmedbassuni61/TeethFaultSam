```python
import plotly.graph_objects as go
import trimesh
import numpy as np
import matplotlib.pyplot as plt

# 1. Define paths to your original mesh and the predicted labels
mesh_path = '/content/TeethFaultSam/raw_data/upper/case_001/case_001_upper.obj'
label_path = '/content/TeethFaultSam/inference_results/labels/upper/case_001_upper.txt'

# 2. Load the mesh and the vertex labels
mesh = trimesh.load(mesh_path)
labels = np.loadtxt(label_path, dtype=int)

# 3. Create a distinct color map for the segments
unique_labels = np.unique(labels)
colors = np.zeros((len(labels), 3))
cmap = plt.get_cmap('tab20')

for lbl in unique_labels:
    if lbl == 0:
        # Give background (class 0, usually gums) a neutral pinkish-gray
        color = [200, 150, 150]
    else:
        # Assign distinct colors to different teeth
        color = np.array(cmap(lbl % 20)[:3]) * 255

    # Apply color to all vertices belonging to this label
    colors[labels == lbl] = color

# Plotly expects RGB strings like 'rgb(255, 0, 0)'
vertex_colors = [f'rgb({int(c[0])}, {int(c[1])}, {int(c[2])})' for c in colors]

# 4. Render the interactive 3D plot
fig = go.Figure(data=[go.Mesh3d(
    x=mesh.vertices[:, 0],
    y=mesh.vertices[:, 1],
    z=mesh.vertices[:, 2],
    i=mesh.faces[:, 0],
    j=mesh.faces[:, 1],
    k=mesh.faces[:, 2],
    vertexcolor=vertex_colors,
    showscale=False
)])

fig.update_layout(
    title="TeethFaultSam Segmentation Output",
    scene=dict(
        xaxis=dict(visible=False), # Hide grid lines for a cleaner look
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
        aspectmode='data'          # Keep 1:1:1 aspect ratio so it doesn't stretch
    ),
    margin=dict(l=0, r=0, b=40, t=40)
)

fig.show()
```
